```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("supermarket_sales_dataset.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
print("Number of rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python

print("Column names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
total_sales = df["Sales_KES"].sum()

print("Total sales revenue in KES:", total_sales)
```

```python
average_sales = df["Sales_KES"].mean()

print("Average sales value per transaction in KES:", average_sales)
```

```python
highest_sales = df["Sales_KES"].max()

print("Highest sales value in KES:", highest_sales)
```

```python
highest_transaction = df.loc[df["Sales_KES"].idxmax()]
print("Details of the transaction with the highest sales value:")
print(highest_transaction)
```

```python
branch_sales = df.groupby("Branch")["Sales_KES"].sum()

print("Total sales by branch:")
print(branch_sales)
```

```python
highest_branch = branch_sales.idxmax()
highest_branch_sales = branch_sales.max()

print("Highest-performing branch:", highest_branch)
print("Total sales:", highest_branch_sales, "KES")
```

```python
category_sales = df.groupby("Category")["Sales_KES"].sum()

print("Total sales by product category:")
print(category_sales)
```

```python
highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

print("Highest-performing category:", highest_category)
print("Total sales:", highest_category_sales, "KES")
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Sales_KES", bins=20, kde=True)

plt.title("Distribution of Sales Revenue in KES")
plt.xlabel("Sales (KES)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (KES)")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
